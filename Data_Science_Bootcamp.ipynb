1. What is the difference between a list, a dictionary, and a NumPy array? Give one example of each.

Lists (list)
   Type: Ordered, mutable sequence.

   Indexing: Integer-based

   Elements: Can mix types

Dictionaries (dict)

    Type: Unordered, mutable key-value pairs.

    Indexing: Uses keys (any hashable type, e.g., strings, numbers).

    Elements: {key: value} pairs (keys must be unique).

NumPy Arrays (numpy.ndarray)

    Type: Fixed-size, homogeneous (all elements same type).

    Indexing: Integer-based (like lists), but optimized for math.

    Elements: Must be same type (int, float, etc.).

4. What does .shape and .describe() tell you about a pandas DataFrame? Show using an example.

1, .shape

    Purpose: Returns the dimensions of the DataFrame as a tuple (rows, columns).
    use : Before merging DataFrames (check if dimensions align).

    After filtering data (verify row count changes).

2, .describe()

    Purpose: Generates summary statistics for numerical columns.
    use : Exploratory Data Analysis (EDA) to spot anomalies (e.g., min salary $0?).

    Compare distributions across datasets.

Example for .shape and .describe

```python
import pandas as pd

# Sample DataFrame
data = {
    'Age': [25, 30, 22, 35, 28],
    'Salary': [50000, 60000, 48000, 80000, 52000]
}

df = pd.DataFrame(data)

# Check shape
print("Shape of DataFrame:", df.shape)

# Get summary statistics
print("\nDescriptive Stats:\n", df.describe())
```

10. What does it mean if a column has a high standard deviation?

When a column has a high standard deviation, it indicates that the values are widely spread out from the average (mean). This suggests greater variability in the data.

In contrast, a low standard deviation means the values are tightly grouped around the mean — indicating more consistency and predictability.

      Create a list or column that has a low and a high std. dev, and compare.  

```python
import numpy as np

low_std_list = [50, 51, 49, 50, 52]
high_std_list = [10, 90, 30, 70, 50]

low_mean = np.mean(low_std_list)
low_std = np.std(low_std_list)

high_mean = np.mean(high_std_list)
high_std = np.std(high_std_list)

print("Low Std. Dev. List:")
print("Values:", low_std_list)
print("Mean:", low_mean)
print("Standard Deviation:", round(low_std, 2))
print("-> These values are close to the mean. Less variation.\n")

print("High Std. Dev. List:")
print("Values:", high_std_list)
print("Mean:", high_mean)
print("Standard Deviation:", round(high_std, 2))
print("-> These values are spread far from the mean. More variation.\n")


import matplotlib.pyplot as plt

plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.plot(low_std_list, marker='o')
plt.title("Low Std Dev")
plt.axhline(y=low_mean, color='r', linestyle='--', label='Mean')
plt.legend()


plt.subplot(1, 2, 2)
plt.plot(high_std_list, marker='o', color='orange')
plt.title("High Std Dev")
plt.axhline(y=high_mean, color='r', linestyle='--', label='Mean')
plt.legend()

plt.tight_layout()
plt.show()
```
